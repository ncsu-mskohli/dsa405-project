# 1. The Audit

````python
# ---------------------------------------------------------------------------
# P2 setup
# ---------------------------------------------------------------------------

import pandas as pd, numpy as np

DATA = "https://raw.githubusercontent.com/ncsu-mskohli/dsa405-project/main/data/raw/"
df = pd.read_csv(DATA + "leith_used_cars.csv")
````

````python
# CLEANING CODEBLOCK 1

# The original dataset, referenced as `df` in this file...
# ...has the information of multiple variables...
# ...contained in just one: "Make & Model"...
# ...and so we must first separate them.

# note that pandas automatically removed the first row with the headers, the csv had 1992 rows

pattern_condition = r'^(Used|Certified Pre-Owned)'
df['Condition'] = df['Make & Model'].str.extract(pattern_condition, expand=False)

temp_series = df['Make & Model'].str.replace(pattern_condition, '', regex=True).str.strip()

df['Year'] = temp_series.str.extract(r'^(\d{4})', expand=False)

df['Make'] = temp_series.str.extract(r'^\d{4}\s+(\w+)', expand=False)

df['Model'] = temp_series.str.replace(r'^\d{4}\s+\w+\s+', '', regex=True).str.strip()

df_c = df[['Condition', 'Year', 'Make', 'Model', 'Price']].copy()

# Printing a sample of our reorganized dataframe, `df_c`
firstotherCondition = (df_c[df_c['Condition'] == 'Certified Pre-Owned'].index[0])
print(df_c.head(1 + firstotherCondition + 1))
````

````python
# Dataframe info
df_c.info()
````

````python
# Auditing the `Condition` column

# Reading it in as the dtype it should be
Condition_audit = (df_c.Condition).astype('string')

# Checking for null values
print(f"Null values in the original: {df_c.Condition.isna().sum()}")
print(f"Null values after reading it as `string`: {Condition_audit.isna().sum()}")
print()

# The number of distinct values and...
#...outputting their full set if it is relevant and reasonable.
print(f"Distinct values in the original: {df_c.Condition.nunique()}")
print(f"Distinct values after reading it in as strings: {Condition_audit.nunique()}")
print()
print(Condition_audit.unique())
print()

# Frequency of values
print(Condition_audit.value_counts())
````

````python
# Auditing the `Year` column

# Reading it in as the dtype it should be
Year_audit = pd.to_numeric(df_c['Year'], errors='coerce')

# Checking for null values
print(f"Null values in the original: {df_c.Year.isna().sum()}")
print(f"Null values after reading it as numeric: {Year_audit.isna().sum()}")
print()

# The number of distinct values and...
#...outputting their full set if it is relevant and reasonable.
print(f"Distinct values in the original: {df_c.Year.nunique()}")
print(f"Distinct values after reading it in as numeric: {Year_audit.nunique()}")
print()
print("Sorted full set of values, ascending:")
print(np.sort(Year_audit.unique()))
print()

# Max, min, range
Year_max = Year_audit.max()
Year_min = Year_audit.min()
Year_range = Year_max - Year_min
print(f"Max: {Year_max} Min: {Year_min} Range: {Year_range}")
print()

# The frequency of each value
print(Year_audit.value_counts(dropna=False))
````

````python
# Auditing the `Make` column

# Reading it in as the dtype it should be
Make_audit = (df_c.Make).astype('string')

# Checking for null values
print(f"Null values in the original: {df_c.Make.isna().sum()}")
print(f"Null values after reading it as `string`: {Make_audit.isna().sum()}")
print()

# The number of distinct values and...
#...outputting their full set if it is relevant and reasonable.
print(f"Distinct values in the original: {df_c.Make.nunique()}")
print(f"Distinct values after reading it in as strings: {Make_audit.nunique()}")
print()
print(Make_audit.unique())
print()

# Frequency of top values
print(Make_audit.value_counts().head(10))
````

````python
# Auditing the `Model` column

# Reading it in as the dtype it should be
Model_audit = (df_c.Model).astype('string')

# Checking for null values
print(f"Null values in the original: {df_c.Model.isna().sum()}")
print(f"Null values after reading it as `string`: {Model_audit.isna().sum()}")
print()

# The number of distinct values and...
#...outputting their full set if it is relevant and reasonable.
print(f"Distinct values in the original: {df_c.Model.nunique()}")
print(f"Distinct values after reading it in as strings: {Model_audit.nunique()}")
print()

# Frequency of top values
print(Model_audit.value_counts().head(20))

# Checking for whitespace causing duplicates
print()
Model_audit_nunique_beforewhitespacestrip = Model_audit.nunique()
Model_audit_nunique_afterwhitespacestrip = Model_audit.str.strip().nunique()
assert Model_audit_nunique_beforewhitespacestrip == Model_audit_nunique_afterwhitespacestrip
````

````python
# Auditing the `Price` column

# Reading it in as the dtype it should be
Price_audit = pd.to_numeric(df_c['Price'], errors='coerce')

# Checking for null values
print(f"Null values in the original: {df_c.Price.isna().sum()}")
print(f"Null values after reading it as numeric: {Price_audit.isna().sum()}")
print()

# Checking for null values after stripping the `$`
Price_audit = df_c['Price'].str.replace(r'[\$,]', '', regex=True)
Price_audit = pd.to_numeric(Price_audit, errors='coerce')
print(f"Null values after stripping the `$`: {Price_audit.isna().sum()}")
print(f"Rate (rounded to 1 decimal place): {Price_audit.isna().sum() / len(Price_audit) * 100:.1f}%")
print()

# The number of distinct values and...
#...outputting their full set if it is relevant and reasonable.
print(f"Distinct values in the original: {df_c.Price.nunique(dropna=False)}")
print(f"Distinct values after reading it in as numeric: {Price_audit.nunique(dropna=False)}")
print(f"Distinct values after reading it in as numeric, non-NaN: {Price_audit.nunique()}")
print()

# Max, min, and range
Price_max = Price_audit.max()
Price_min = Price_audit.min()
Price_range = Price_max - Price_min
print(f"Max: {Price_max} Min: {Price_min} Range: {Price_range}")
print()

print(df_c.iloc[Price_audit[Price_audit == Price_max].index[0]])
print()
print(df_c.iloc[Price_audit[Price_audit == Price_min].index[0]])
print()

# Frequency, checking for duplicates due to distinct values not being as high as expected
print(Price_audit.value_counts().head(20))
print()
print(df_c[Price_audit == 19998.0])
````

# 3. The cleaning log
| # | Column | Change made | Rows/cells affected | Why | What is lost |
|---|---|---|---|---|---|
| 1 | `Make & Model` | Four new columns created from it in dataframe `df`, them and `Price` counted into the new dataframe `df_c` | 1991 rows | To separate variables stuck together in one column. | Nothing |
| 2 | `Condition` | Converting from `object` to `category` | 1991 cells | More appropriate data type, efficiency | Nothing |
| 3 | `Make` | Converting from `object` to `category` | 1991 cells | More appropriate data type, efficiency | Nothing |
| 4 | `Model` | Converting from `object` to `string` | 1991 cells | More appropriate data type, efficiency | Nothing |
| 5 | `Price` | Stripping `$` from values and then converting via `pd.to_numeric` | 1991 cells | Identify truly null values; More appropriate data type, efficiency | Nothing |
| 6 | `Price` | Removing rows which have a null value for this column | 38 rows | No longer useful for the dataset, data could not be repaired, it was simply missing | 38 rows |
| 7| `Year` | Converting via `pd.to_numeric` | 1953 cells | More appropriate data type, efficiency | Nothing |

````python
# CLEANING CODEBLOCK 2
# Reseting everything
%reset -f
# Making changes based on the audit
# All decisions recorded in the cleaning log

# Setup again
import pandas as pd, numpy as np

DATA = "https://raw.githubusercontent.com/ncsu-mskohli/dsa405-project/main/data/raw/"
df = pd.read_csv(DATA + "leith_used_cars.csv")

print("Info on the original dataframe")
print("-")
df.info(memory_usage='deep')
print()

# 1
pattern_condition = r'^(Used|Certified Pre-Owned)'
df['Condition'] = df['Make & Model'].str.extract(pattern_condition, expand=False)

temp_series = df['Make & Model'].str.replace(pattern_condition, '', regex=True).str.strip()

df['Year'] = temp_series.str.extract(r'^(\d{4})', expand=False)

df['Make'] = temp_series.str.extract(r'^\d{4}\s+(\w+)', expand=False)

df['Model'] = temp_series.str.replace(r'^\d{4}\s+\w+\s+', '', regex=True).str.strip()

df_c = df[['Condition', 'Year', 'Make', 'Model', 'Price']].copy()

print("Information after column reorganization prior to cleaning")
print("-")
df_c.info(memory_usage='deep')
print()

# 2
prior = df_c.Condition.nunique(dropna=False)
df_c['Condition'] = df_c['Condition'].astype('category')
after = df_c.Condition.nunique(dropna=False)
assert prior == after

# 3
prior = df_c.Make.nunique(dropna=False)
df_c['Make'] = df_c['Make'].astype('category')
after = df_c.Make.nunique(dropna=False)
assert prior == after

# 4
prior = df_c.Model.nunique(dropna=False)
df_c['Model'] = df_c['Model'].astype('string')
after = df_c.Model.nunique(dropna=False)
assert prior == after

# 5
Price_nulls = df_c.Price.isna().sum()
df_c['Price'] = df_c['Price'].str.replace(r'[\$,]', '', regex=True)
df_c['Price'] = pd.to_numeric(df_c['Price'], errors='coerce')
Price_nulls = df_c.Price.isna().sum()

# Added during step 6 as a pre-check, consider this "5.5"
# Was originally `assert` and not `assert not`
assert not len(df_c) == len(df_c.dropna())

# 6
priorRows = len(df_c)
df_c = df_c.dropna(subset=['Price'])
postRows = len(df_c)
assert df_c.Price.isna().sum() == 0
assert postRows == priorRows - Price_nulls

# 7
prior = df_c.Year.nunique(dropna=False)
df_c['Year'] = pd.to_numeric(df_c['Year'], errors='coerce')
after = df_c.Year.nunique(dropna=False)
assert prior == after

# This confirms no more nulls exist as len counts NaN values
assert len(df_c) == len(df_c.dropna())

# Info after cleaning
print("Information after cleaning")
print("-")
df_c.info(memory_usage='deep')
print()
print(df_c.nunique())

# Saving the cleaned dataset into the temporary local directory, "files"
df_c.to_csv('leith_used_cars_cleanedX.csv', index=False)
````

# 5. The Provenance Brief
Leith Cars is a car dealership in Raleigh, NC. They maintain an inventory of used cars they have for sale on their website, `leithcars.com`.

Through a scrape of the website on 10/03/2026, we gathered the following information on each car: its `Condition`, whether it is simply `Used` or also `Certified Pre-Owned`; its `Year` of manufacture; its `Make`; its `Model`; and finally, and most importantly, its `Price`.

Mileage was not scraped although it was information given on the website. We will rely on the year of manufacture to serve its purpose as an indicator.

We then organized and cleaned this data to end up with a dataset that lists 1953 used cars listed for sale on October 3rd, 2026 that were manufactured between 2002 and 2027, of 39 makes, and 975 models.

The vast majority of these cars, nearly 80% of them, were manufactured between `2019` and `2025`. The makes are more evenly distributed. The largest share is of `Honda`, which accounts for 15% of the cars being sold.

We intend to use this dataset to help make comparisons between Raleigh's used car market with national data.
